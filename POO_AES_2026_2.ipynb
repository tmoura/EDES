# **Exemplos de POO - Algoritmos e Estrutura de Dados - 2026.2**

```python
class Ponto:
  def __init__(self, x, y):
    self.x = x
    self.y = y


if __name__ == "__main__":
  p1 = Ponto(10, 20)
  p2 = Ponto(30, 40)

  print(f"Ponto 1 tem dimensões: {p1.x}, {p1.y}")
  print(f"Ponto 2 tem dimensões: {p2.x}, {p2.y}")
```

```python
class Retangulo:
  def __init__(self,base, altura):

    self.base = base
    self.altura = altura


if __name__ == "__main__":
  r1 = Retangulo(10, 20)

  print(f"Área do Retângulo é: {r1.base * r1.altura}")
```

```python
class Aluno:
  def __init__(self, nome, matricula):
    self.nome = nome
    self.matricula = matricula

if __name__ == "__main__":
  aluno1 = Aluno("João", "12345")
  aluno2 = Aluno("Maria", "67890")

  print(f"Aluno 1: {aluno1.nome}, Matrícula: {aluno1.matricula}")
  print(f"Aluno 2: {aluno2.nome}, Matrícula: {aluno2.matricula}")
```

```python
class Conta_Corrente:
  def __init__(self, numero, saldo):
    self.numero = numero
    self.saldo = saldo

if __name__ == "__main__":
  conta1 = Conta_Corrente("12345", 1000.0)
  conta2 = Conta_Corrente("67890", 500.0)

  print(f"Conta 1: Número {conta1.numero}, Saldo {conta1.saldo}")
  print(f"Conta 2: Número {conta2.numero}, Saldo {conta2.saldo}")
#
```

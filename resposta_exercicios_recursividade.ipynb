## Questão 1:

```python
def exponencial(X, N):
  if N == 0:
    return 1
  else:
    return X * exponencial(X, N - 1)


if __name__ == '__main__':

  print(exponencial(2, 10))
```

## Questão 2:

```python
def fib(n):
  if n == 1:
    return 0
  elif n == 2:
    return 1
  else:
    return fib(n - 1) + fib(n - 2)


if __name__ == '__main__':

  print(fib(12))
```

## Questão 3:

```python
class No:
  def __init__(self, dado = None):
    self.__dado = dado
    self.__prox = None

  # get
  @property
  def dado(self):
    return self.__dado

  # set
  @dado.setter
  def dado(self, novo):
    self.__dado = novo

  # get
  @property
  def prox(self):
    return self.__prox

  # set
  @prox.setter
  def prox(self, novo):
    self.__prox = novo
```

```python
class Lista_Encadeada_v2:

  def __init__(self):
    self.__inicio = None

  @property
  def inicio(self):
    return self.__inicio

  # Se pos for valor um, zero ou negativo, colocar o no na primeira posição
  # Se pos tiver um valor maior do que o tamanho da lista, colocar na última posição
  def inserir(self, pos, no):
    if self.__inicio == None:
      self.__inicio = no
    elif pos <= 1:
      no.prox = self.__inicio
      self.__inicio = no
    elif pos >= 2 and pos <= self.tamanho():
      p = self.__inicio
      q = None
      i = 1

      while i < pos:
        q = p
        p = p.prox
        i += 1

      q.prox = no
      no.prox = p
    elif pos > self.tamanho():
      p = self.__inicio

      while p.prox != None:
        p = p.prox

      p.prox = no


  # Se pos for valor um, zero ou negativo, remover o no na primeira posição
  # Se pos tiver um valor maior do que o tamanho da lista, remover da última posição
  def remover(self, pos):
    if self.__inicio == None:
      return None
    elif pos <= 1:
      p = self.__inicio
      self.__inicio = self.__inicio.prox
      return p.dado
    elif pos >= 2 and pos <= self.tamanho():
      p = self.__inicio
      q = None
      i = 1

      while i < pos:
        q = p
        p = p.prox
        i += 1

      q.prox = p.prox
      return p.dado # dado retornado

    elif pos > self.tamanho():
      p = self.__inicio
      q = None

      while p.prox != None:
        q = p
        p = p.prox

      q.prox = None
      return p.dado # último elemento retornado

  # Se pos for valor zero ou negativo, consultar o no na primeira posição
  # Se pos tiver um valor maior do que o tamanho da lista, consultar da última posição
  def consultar(self, pos):
    if self.__inicio == None:
      return None
    elif pos <= 1:
      self.__inicio.dado
    elif pos >= 2 and pos <= self.tamanho():
      p = self.__inicio
      q = None
      i = 1

      while i < pos:
        p = p.prox
        i += 1

        return p.dado
    elif pos > self.tamanho():
      p = self.__inicio

      while p.prox != None:
        p = p.prox

      return p.dado

  def tamanho(self):
    p = self.__inicio

    tamanho = 0
    while p != None:
      tamanho += 1
      p = p.prox
    return tamanho

  def __str__(self):
    saida = 'Lista_v2: ['
    p = self.__inicio

    while p != None:
      saida += f'{p.dado}'
      p = p.prox

      if p != None:
        saida += ', '

    saida += ']'
    return saida
```

## Questão 4:

```python
def quantidade_brancos(palavra):
  if palavra == '':
    return 0
  else:
    if palavra[0] == ' ':
      return 1 + quantidade_brancos(palavra[1:])
    else:
      return quantidade_brancos(palavra[1:])


if __name__ == "__main__":

  texto = "Alo Mundo! Tudo bem?"

  print(quantidade_brancos(texto))
```
